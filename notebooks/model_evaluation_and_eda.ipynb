# CardioGuard AI - Exploratory Data Analysis & Machine Learning Evaluation
### Comprehensive Analysis of the UCI Cleveland Heart Disease Dataset
This notebook documents:
1. Dataset loading and verification
2. Exploratory Data Analysis (EDA) and distribution checks
3. Missing value imputation and categorical encoding strategy
4. Multi-model training and cross-validation comparison
5. Metrics: Accuracy, Precision, Recall/Sensitivity, Specificity, F1, ROC-AUC
6. Explainability and feature attribution using SHAP

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import json

data_path = '../dataset/heart_disease_cleaned.csv'
df = pd.read_csv(data_path)
print('Dataset Shape:', df.shape)
display(df.head())
print('Target Distribution:')
print(df['target'].value_counts(normalize=True))
```

## 2. Demographic and Clinical Feature Distribution

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
df['age'].hist(ax=axes[0, 0], bins=15, color='#1e3a8a', edgecolor='white')
axes[0, 0].set_title('Age Distribution')

df['trestbps'].hist(ax=axes[0, 1], bins=15, color='#0284c7', edgecolor='white')
axes[0, 1].set_title('Resting Blood Pressure')

df['chol'].hist(ax=axes[0, 2], bins=15, color='#0d9488', edgecolor='white')
axes[0, 2].set_title('Serum Cholesterol')

df['thalach'].hist(ax=axes[1, 0], bins=15, color='#4f46e5', edgecolor='white')
axes[1, 0].set_title('Max Heart Rate (thalach)')

df['oldpeak'].hist(ax=axes[1, 1], bins=15, color='#dc2626', edgecolor='white')
axes[1, 1].set_title('ST Depression (oldpeak)')

df['target'].value_counts().plot(kind='bar', ax=axes[1, 2], color=['#10b981', '#f43f5e'])
axes[1, 2].set_title('Target Class Balance (0: Normal, 1: Disease)')
axes[1, 2].set_xticklabels(['Normal', 'Heart Disease'], rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Load Trained Model Comparison & Evaluation Artifacts

```python
with open('../backend/saved_models/model_metadata.json', 'r') as f:
    meta = json.load(f)

comp_df = pd.DataFrame(meta['model_comparison'])
display(comp_df[['model_name', 'cv_roc_auc_mean', 'test_accuracy', 'test_recall', 'test_specificity', 'test_f1', 'test_roc_auc']])
```

## 4. Feature Importance & SHAP Summary

```python
feat_df = pd.DataFrame(meta['feature_importance'])
plt.figure(figsize=(10, 6))
plt.barh(feat_df['label'], feat_df['importance'], color='#0ea5e9')
plt.gca().invert_yaxis()
plt.xlabel('Relative Feature Importance (%)')
plt.title('Global Feature Importance Attribution - CardioGuard AI')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
